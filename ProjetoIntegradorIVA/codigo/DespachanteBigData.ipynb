```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Despachante Globo") \
    .getOrCreate()
```

```python
clientes = spark.read.csv("clientes.csv", header=True, inferSchema=True)
servicos = spark.read.csv("servicos.csv", header=True, inferSchema=True)
atendimentos = spark.read.csv("atendimentos.csv", header=True, inferSchema=True)
```

```python
atendimentos.groupBy().sum("valor_final").show()
```

```python
atendimentos.groupBy("id_servico").count().show()
```

```python
atendimentos.groupBy().avg("tempo_dias").show()
```

```python
df = atendimentos.join(clientes, "id_cliente") \
                 .join(servicos, "id_servico")

df.show()
```
